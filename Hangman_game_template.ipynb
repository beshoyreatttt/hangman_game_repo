## Description
> ##### The Hangman program randomly selects a secret word from a list of secret words.
> ##### The Game: Here, a random word (a fruit name) is picked up from our collection and the player gets limited chances to guess the fruit name correctly.
> ##### When a letter in that word is guessed correctly, that letter position in the word is made visible in the output.
> ##### In this way, all letters of the word are to be guessed before all the chances are over.
> ##### For convenience, the number of chances equals the word length + 2.
> ##### For example, word to be guessed is mango, then user gets 5 + 2 = 7 chances, as mango is a five letter word.
> ##### Note : Total number of Guesses either right or wrong equals word length + 2.
> ##### Make game repeats until player ends the game.
> ##### Bonus : Create a function that takes 3 parameters `Easy`, `Medium` and `Hard` where `Medium` has more difficult words to be guessed than `Easy` and so on.

```python

import random


def choose_level():
    level = input("Choose Level (Easy / Medium / Hard): ").capitalize()

    while level not in ["Easy", "Medium", "Hard"]:
        print("Invalid Level! Try again.")
        level = input("Choose Level (Easy / Medium / Hard): ").capitalize()

    return level


def choose_word(level):
    easy = ["apple", "mango", "grape", "lemon"]
    medium = ["cherry", "coconut", "avocado", "apricot"]
    hard = ["strawberry", "pineapple", "watermelon", "blueberry"]

    if level == "Easy":
        word = random.choice(easy)

    elif level == "Medium":
        word = random.choice(medium)

    else:
        word = random.choice(hard)

    return word


def display_word(word, guessed_letters):
    result = ""

    for letter in word:
        if letter in guessed_letters:
            result += letter + " "
        else:
            result += "_ "

    print("\nFruit:", result)


def get_letter(guessed_letters):
    letter = input("Guess a letter: ").lower()

    while len(letter) != 1 or not letter.isalpha():
        print("Please enter one letter only!")
        letter = input("Guess a letter: ").lower()

    while letter in guessed_letters:
        print("You already guessed this letter!")
        letter = input("Guess another letter: ").lower()

        while len(letter) != 1 or not letter.isalpha():
            print("Please enter one letter only!")
            letter = input("Guess a letter: ").lower()

    return letter


def check_win(word, guessed_letters):
    for letter in word:
        if letter not in guessed_letters:
            return False

    return True


def play_game():
    print("\nWelcome to Hangman Game!")
    print("------------------------")

    level = choose_level()

    word = choose_word(level)

    guesses = len(word) + 2

    guessed_letters = []

    print("\nLevel:", level)
    print("You have", guesses, "guesses.")

    while guesses > 0:

        display_word(word, guessed_letters)

        print("Guesses left:", guesses)

        letter = get_letter(guessed_letters)

        guessed_letters.append(letter)

        if letter in word:
            print("Correct Guess!")

        else:
            print("Wrong Guess!")
        
        guesses -= 1

        if check_win(word, guessed_letters):
            display_word(word, guessed_letters)
            print("\nCongratulations! You Win!")
            print("The fruit is:", word)
            break

    if guesses == 0 and not check_win(word, guessed_letters):
        print("\nGame Over!")
        print("The fruit was:", word)


def play_again():
    answer = input("\nDo you want to play again? (yes/no): ").lower()

    while answer not in ["yes", "no"]:
        print("Please enter yes or no.")
        answer = input("Do you want to play again? (yes/no): ").lower()

    return answer


answer = "yes"

while answer == "yes":

    play_game()

    answer = play_again()

print("\nThanks for playing Hangman!")


```
